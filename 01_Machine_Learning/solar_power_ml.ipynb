```python
import os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from train_and_save import generate_solar_dataset, run_ml_pipeline

# Set plot styles
sns.set_theme(style="darkgrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

## 1. Data Collection & Inspection

```python
df = generate_solar_dataset(n_samples=1500)
print("Dataset Shape:", df.shape)
df.head()
```

## 2. Exploratory Data Analysis (EDA)

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Numerical Distributions
df.hist(figsize=(14, 10), bins=25, color='skyblue', edgecolor='black')
plt.tight_layout()
plt.show()
```

## 3. Run Full ML Pipeline & Benchmark Models

```python
# Run training pipeline script to fit all models
run_ml_pipeline()
```

## 4. Evaluation Plots & Comparison

```python
reg_df = pd.read_csv('models/regression_metrics.csv')
cls_df = pd.read_csv('models/classification_metrics.csv')

# Regression Comparison Bar Chart
plt.figure(figsize=(12, 5))
sns.barplot(data=reg_df, x='Model', y='R2 Score', palette='viridis')
plt.title('Regression Model Comparison (R2 Score)')
plt.xticks(rotation=45)
plt.ylim(0, 1)
plt.show()

# Classification Comparison Bar Chart
plt.figure(figsize=(12, 5))
sns.barplot(data=cls_df, x='Model', y='F1 Score', palette='magma')
plt.title('Classification Model Comparison (Weighted F1 Score)')
plt.xticks(rotation=45)
plt.ylim(0, 1)
plt.show()
```
